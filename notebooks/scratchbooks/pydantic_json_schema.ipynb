```python
from pydantic import BaseModel
```

```python
from timeit import timeit

from pydantic import TypeAdapter
from typing_extensions import TypedDict


class A(TypedDict):
    a: str
    b: int


class TypedModel(TypedDict):
    a: A


class B(BaseModel):
    a: str
    b: int


class Model(BaseModel):
    b: B


ta = TypeAdapter(TypedModel)
result1 = timeit(lambda: ta.validate_python({"a": {"a": "a", "b": 2}}), number=10000)
result2 = timeit(lambda: Model.model_validate({"b": {"a": "a", "b": 2}}), number=10000)
print(result2 / result1)
```

```python
from pydantic._internal._core_utils import pretty_print_core_schema

pretty_print_core_schema(ta.json_schema())
```

```python
pretty_print_core_schema(Model.model_json_schema())
```
