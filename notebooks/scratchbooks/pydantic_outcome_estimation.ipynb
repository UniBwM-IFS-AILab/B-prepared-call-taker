```python
from ems_prepared.dialogue_state.type_defs import KnownBoolean, Unknown
```

```python
cases = (
    [Unknown, Unknown],  # -> None
    [Unknown, False],  # -> None
    [Unknown, True],  # -> True
    [True, False],  # -> True
    [True, True],  # -> True
    [False, False],  # -> False
    [Unknown, False, True],
)  # -> True
```

```python
any(cases[0])  # needs to return None
```

```python
def tristate(values) -> KnownBoolean:
    from functools import reduce

    def tri_combine(acc, curr):
        # single True is sufficient
        if acc is True or curr is True:
            return True
        # All needs to be False
        if acc is False and curr is False:
            return False
        # at least one None, nothing True yet
        return Unknown

    return reduce(tri_combine, values, Unknown)
```

```python
print(tristate(cases[0]))
```

```python
[tristate(case) for case in cases]
```
