```python
from deepmerge import Merger
from ems_prepared.state_model.custom_deepmerge import ignore_empty_merger

# 1️⃣ Build a merger
skip_empty_merger = Merger(
    type_strategies=[
        (dict, ["merge"]),
        (list, ["append"]),
    ],
    fallback_strategies=["override"],
    type_conflict_strategies=["override_if_not_empty"],
)

# 2️⃣ Example data
base = {
    "name": "Alice",
    "age": 42,
    "tags": ["v1"],
    "profile": {"bio": "Loves Python 🐍", "chink": "shit", "a": None, "c": None},
}

patch = {
    "name": "",  # empty string - should NOT clobber “Alice”
    "age": None,  # None - should NOT clobber 42
    "tags": [],  # empty list - ignored, existing list kept
    "profile": {"bio": None, "a": "a", "b": None},  # nested empty - ignored
}

# 3️⃣ Merge in-place (deepmerge always mutates the first arg)
ignore_empty_merger.merge(base, patch)

print(base)
# ➜ {'name': 'Alice',
#     'age': 42,
#     'tags': ['v1'],
#     'profile': {'bio': 'Loves Python 🐍'}}
```

```python
from collections import ChainMap

context = ChainMap(base, patch)
print(dict(context))
```

```python
import sys

sys.path.append(r"e:\work\ems-prepared\src")

from rich import print

from ems_prepared.state_model.emergency_call_state import EmergencyCall
from ems_prepared.state_model.medical_symptoms_state import MedicalEmergency
from ems_prepared.state_model.type_defs import EmergencyType
from ems_prepared.state_model.type_defs import Unknown

# 🚨 COMPLEX EXAMPLE 1: Partial call data with empty values

# Base call data (initial incomplete information)
base_call = EmergencyCall(
    caller_name="",  # empty string - caller didn't provide name initially
    caller_phone="555-0123",
    location=None,  # location not determined yet
    emergency_type=EmergencyType.MEDICAL,
    situation_description="Person collapsed",
    medical_emergency=MedicalEmergency(
        # Breathing symptoms
        severe_dyspnoea=True,
        now_sufficient_air=None,  # not assessed yet
        # Consciousness - mixed data
        unconscious=False,
        now_responsive_normally=None,  # not checked
        # Empty nested structures
        acute_chest_pain=None,
        heavy_non_traumatic_bleeding=None,
    ),
)

# Update call data (more information gathered)
update_call = EmergencyCall(
    caller_name="John Smith",  # now we have the name
    caller_phone="",  # empty string - should not override existing
    location="123 Main St",  # location now available
    emergency_type=None,  # no change to emergency type
    situation_description="",  # empty - should not override
    medical_emergency=MedicalEmergency(
        # New breathing info
        severe_dyspnoea=None,  # no change
        now_sufficient_air=False,  # new info
        # Consciousness updates
        unconscious=Unknown,  # no change
        now_responsive_normally=False,  # new assessment
        # New symptoms
        acute_chest_pain=True,
        heavy_non_traumatic_bleeding=False,
    ),
)

print("🔵 EXAMPLE 1: Partial call updates")
print(base_call)
print()

print("Update call:")
print(update_call)
```

```python
# 🔗 ChainMap approach for Example 1
print("\n🔗 CHAINMAP RESULT:")
print("-" * 30)

# Convert to dicts for ChainMap
base_dict = base_call.model_dump()
update_dict = update_call.model_dump()

# ChainMap prioritizes first non-None value
chain_result = ChainMap(update_dict, base_dict)
merged_dict = dict(chain_result)

print(merged_dict)
```

```python
# 🔄 Deepmerge approach for Example 1
print("\n🔄 DEEPMERGE RESULT:")
print("-" * 30)

# Make copies to avoid mutating originals
base_copy = base_call.model_copy(deep=True).model_dump()
update_copy = update_call.model_dump()

# Apply deepmerge with skip_empty strategy
skip_empty_merger.merge(base_copy, update_copy)

print(base_copy)
```
